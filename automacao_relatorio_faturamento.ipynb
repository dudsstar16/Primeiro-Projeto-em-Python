### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# Bibliotecas a serem baixadas no terminal 
# para instalar no terminal: pip install pandas pyautogui openpyxl (prompt cmd)
# pyautogui -> automações (coontrola o mouse e o teclado do pc, robo) 
# pandas -> dados 
# openpyxl -> (abrir base de dados em excel .xlsx)
```

```python
# importar as bibliotecas para inciar o projeto 

import pyautogui
import time

pyautogui.PAUSE = 0.5

# Passo a Passo do desafio 

# Passo 1: Entrar no sistema da empresa (link do drive)

pyautogui.press ("win")
pyautogui.write ("chrome")
pyautogui.press ("enter")

time.sleep(1)

# criar variavel de link para nao ter que substituir uma por uma manual caso o link atualize 

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

pyautogui.write (link)

pyautogui.press ("enter")

time.sleep(3)
```

```python
# Passo 2: Navegar no sistema para encontrar a base de dados 

# posição do mouse e quantidade de clicks

pyautogui.click (x=333, y=447, clicks=2)

# Passo 3: Exportar a base de dados (dowload)

time.sleep (1)

pyautogui.click (x=1259, y=447, clicks=1)

time.sleep (1)

pyautogui.click (x=739, y=450, clicks=1)

time.sleep (1)

pyautogui.click (x=1262, y=446, clicks=1)

time.sleep (1)

pyautogui.click (x=1084, y=402, clicks=1)

```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
# importar biblioteca para tratar dados 

import pandas 
import openpyxl

# Passo 4: Calcular os indicadores faturamento e quantidade de produtos 

# abrir a base de dados 

# r na frente le o texto na forma crua, pois \n é uma tecla de enter no python

caminho = r"C:\Users\eduar\Downloads\Vendas - Dez.xlsx"

# pois \ é um caractere de escape, o python entende que o \ é para escapar o próximo caractere, por isso tem que colocar r para ler o texto como crua e não interpretar os \ como caracteres de escape
# r ler o texto de forma crua 

# comando para extrair o dado da tabela 

tabela = pandas.read_excel(caminho)

# ver as infos da tabela 

# ver a tabela com print display e print (diplay vem melhor estruturada)

display(tabela)

# somar o faturamento de todos os produtos 

faturamento = tabela["Valor Final"].sum()

# somar a quantidade de produtos vendidos

qtd_produtos = tabela["Quantidade"].sum()


print(faturamento)
print(qtd_produtos)

```

### Vamos agora enviar um e-mail pelo gmail

```python
# Passo 5: Enviar as informações por  e-mail 


# primeira maneira de fazer 

# pyautogui.press ("win")
# pyautogui.write ("chrome")      
# pyautogui.press ("enter")
# pyautogui.write ("https://mail.google.com/mail/")
# pyautogui.press ("enter")
# time.sleep(3)

# abrir uma nova aba no navegador 

import pyperclip

time.sleep(3)
pyautogui.hotkey ("ctrl", "t")
pyautogui.write ("https://mail.google.com/mail/")
pyautogui.press ("enter")
time.sleep(3)

# clicar para escrever um novo e-mail

pyautogui.click (x=127, y=222, clicks=1)

# preencher os campos do e-mail

pyautogui.write ("eduarda.christina@iesb.edu.br")
pyautogui.press ("tab") 
pyautogui.press ("tab")
time.sleep(2)

pyperclip.copy (f"Relatório de Vendas") # copia o assunto do e-mail para a área de transferência, python nao escreve com acentos, por isso tem que usar o pyperclip para copiar o texto com acentos e colar no e-mail
pyautogui.hotkey ("ctrl", "v") # cola o assunto do e-mail
time.sleep(2)

pyautogui.press ("tab") # seleciona o campo de escrever o e-mail

texto = f"""Prezado(a) equipe,

Segue em anexo o relatório de vendas do mês de dezembro.

O faturamento foi de R${faturamento:,.2f} e a quantidade de produtos vendidos foi de {qtd_produtos:,}.

Atenciosamente,
Eduarda Christina
"""

# o """ é para escrever um texto em várias linhas, o f""" é para formatar o texto com as variáveis dentro do texto, o f
# as chaves {} são para colocar as variáveis dentro do texto, o :,.2f é para formatar o número com vírgula de milhar e 2 casas decimais

pyperclip.copy (texto) # copia o texto do e-mail para a área de transferência
pyautogui.hotkey ("ctrl", "v") # cola o texto do e-mail
time.sleep(2)

# hotkey é para usar as teclas de atalho do teclado, no caso ctrl + v para colar o texto do e-mail

pyautogui.hotkey ("ctrl", "enter") # enviar o e-mail usando o atalho ctrl + enter


```
